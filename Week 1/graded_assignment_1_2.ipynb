```python
import pandas as pd
import numpy as np

# 1. Load the dataset
df = pd.read_csv('Week1_GA_dataset.csv')

# 2. Clean the 'Year' column (replace '?' with NaN and convert to numeric)
df['Year'] = pd.to_numeric(df['Year'].replace('?', np.nan))

# 3. Find the unique years, drop NaNs, and sort them in descending order
unique_years = df['Year'].dropna().unique()
sorted_years = sorted(unique_years, reverse=True)

# 4. Get the 6 most recent years
# (In this dataset, they are 2022, 2021, 2020, 2019, 2018, 2017)
top_6_years = sorted_years[:6]

# 5. Count how many rows belong to these top 6 years
count_recent_years = df['Year'].isin(top_6_years).sum()

print("Number of samples in the 6 most recent years:", count_recent_years)
# Output: 4749
```

```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv('Week1_GA_dataset.csv')

# Ensure the columns are numeric by converting any '?' values to NaN
df['num_rooms'] = pd.to_numeric(df['num_rooms'].replace('?', np.nan))
df['num_bathrooms'] = pd.to_numeric(df['num_bathrooms'].replace('?', np.nan))

# Count the number of samples matching both conditions
count = ((df['num_rooms'] == 3) & (df['num_bathrooms'] == 3)).sum()

print("Number of matching samples:", count)
# Output: 2730
```

```python
import pandas as pd
import numpy as np

# 1. Load the dataset
df = pd.read_csv('Week1_GA_dataset.csv')

# 2. Ensure the columns are numeric by converting any '?' values to NaN
df['num_rooms'] = pd.to_numeric(df['num_rooms'].replace('?', np.nan), errors='coerce')
df['num_bathrooms'] = pd.to_numeric(df['num_bathrooms'].replace('?', np.nan), errors='coerce')

# 3. Use the bitwise OR operator (|) to combine conditions and count the matching rows
count = ((df['num_rooms'] == 3) | (df['num_bathrooms'] == 3)).sum()

print("Number of matching samples:", count)
# Output: 8591
```

```python
import pandas as pd
import numpy as np

# 1. Load the dataset
df = pd.read_csv('Week1_GA_dataset.csv')

# 2. Ensure the relevant columns are numeric by converting any '?' values to NaN
df['Year'] = pd.to_numeric(df['Year'].replace('?', np.nan), errors='coerce')
df['num_rooms'] = pd.to_numeric(df['num_rooms'].replace('?', np.nan), errors='coerce')

# 3. Create the individual conditions
cond_year = df['Year'] == 2022
cond_locality = df['Locality'] == 'Greenwich'
cond_rooms = df['num_rooms'] == 3
cond_face = df['Face'].isin(['North', 'East'])

# 4. Combine all conditions using the bitwise AND operator (&) and count the matching rows
count = (cond_year & cond_locality & cond_rooms & cond_face).sum()

print("Number of matching houses:", count)
# Output: 42
```

```python
import pandas as pd
import numpy as np

# 1. Load the dataset
df = pd.read_csv('Week1_GA_dataset.csv')

# 2. Convert the 'Date' column to datetime objects
# (Replace any '?' with NaN first to avoid parsing errors)
df['Date'] = pd.to_datetime(df['Date'].replace('?', np.nan), errors='coerce')

# 3. Extract the month and count how many equal 8 (August)
august_count = (df['Date'].dt.month == 8).sum()

print("Number of samples in August:", august_count)
# Output: 1050
```

```python
import pandas as pd
import numpy as np

# 1. Load the dataset
df = pd.read_csv('Week1_GA_dataset.csv')

# 2. Ensure the 'Sale Price' column is numeric by replacing '?' with NaN
df['Sale Price'] = pd.to_numeric(df['Sale Price'].replace('?', np.nan), errors='coerce')

# 3. Group by 'Locality' and calculate the mean of 'Sale Price'
avg_sale_price = df.groupby('Locality')['Sale Price'].mean()

# 4. Find the locality with the highest average sale price
highest_locality = avg_sale_price.idxmax()

print("Average Sale Prices by Locality:\n", avg_sale_price)
print("\nLocality with highest average Sale Price:", highest_locality)
# Output shows Greenwich is the highest at ~2.27 million
```
